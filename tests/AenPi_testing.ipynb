```python


from AenPi.experimental import CarbonEstimator
# ---------------- CREATE ESTIMATOR ----------------
carbon = CarbonEstimator()

# ---------------- 1. LLM CARBON ESTIMATION ----------------
llm_result = carbon.estimate_llm(
    llm="gpt-4",
    n_calls=1000,
    avg_tokens=50
)

print("\n=== 🌍 LLM CARBON ESTIMATION ===")
print(llm_result)


# ---------------- 2. REAL FUNCTION MEASUREMENT ----------------
def sample_function():
    return sum(i for i in range(1000))

result = carbon.measure(sample_function, module="sentiment")

print("\n=== ⚡ REAL CPU CARBON MEASUREMENT ===")
print(result)


# ---------------- 3. BATCH TRACKING ----------------
carbon.track(
    module="ner",
    calls=500,
    avg_sec_per_call=0.002
)

carbon.track(
    module="sentiment",
    calls=1000,
    avg_sec_per_call=0.001
)


# ---------------- 4. SESSION SUMMARY ----------------
summary = carbon.summary()

print("\n=== 📊 SESSION CARBON SUMMARY ===")
print(summary)


# ---------------- 5. LLM vs AenPi COMPARISON ----------------
comparison = carbon.compare(
    llm="gpt-3.5",
    module="sentiment",
    n_calls=1000,
    avg_tokens=100
)

print("\n=== 🌱 GREEN AI COMPARISON (LLM vs AenPi) ===")
print(comparison)
```

```python
from AenPi.urdu import UrduSummarizer

# ٹیکسٹ ڈیٹا (آرٹیکل) جس کا خلاصہ تیار کرنا ہے
article_text = """
پاکستان میں انفارمیشن ٹیکنالوجی کے شعبے میں تیزی سے ترقی ہو رہی ہے۔
حکومت نے نوجوانوں کے لیے نئے ڈیجیٹل اور فری لانسنگ پروگرامز کا آغاز کیا ہے۔
اس منصوبے کا بنیادی مقصد ملک بھر میں سافٹ ویئر ڈویلپمنٹ اور نئی ٹیکنالوجی کو فروغ دینا ہے۔
نئی انفراسٹرکچر پالیسی کے تحت آئی ٹی برآمدات میں ریکارڈ اضافہ متوقع ہے۔
ماہرین کا کہنا ہے کہ اس ڈیجیٹل اقدام سے روزگار کے ہزاروں نئے مواقع پیدا ہوں گے اور ملکی معیشت کو استحکام ملے گا۔
"""

# سمرائزر کو لوڈ اور فٹ کرنا
summ = UrduSummarizer()
summ.fit()

print("--- تفصیلی خلاصہ (Scores کے ساتھ) ---")
results = summ.summarize(article_text, n=3)
for r in results:
    print(r["sentence"], "→ score:", r["score"])

print("\n--- فوری ٹیکسٹ خلاصہ (2 جملے) ---")
# Quick string output
summary = summ.summarize_text(article_text, n=2)
print(summary)

print("\n--- اہم الفاظ (Keywords) ---")
# Keywords
keywords = summ.keyword_summary(article_text, top_k=10)
print(keywords)
```
